```python
work = df.copy()

work["date"] = pd.to_datetime(work["date"])

work["price"] = (
    work["price"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.strip()
)

work["price"] = pd.to_numeric(work["price"], errors="coerce")

work["item"] = work["item"].replace({
    "oil b": "Oil B"
})

issues = work[
    work["price"].isna()
    | (work["price"] >= 1000)
]

display(issues)
```

```python

df["date"] = pd.to_datetime(df["date"])

df["price"] = df["price"].astype(str).str.replace(",", "")
df["price"] = pd.to_numeric(df["price"], errors="coerce")


df["item"] = df["item"].replace("oil b", "Oil B")


df = df.dropna(subset=["price"])

df.head()
```

```python
stats = df.groupby("item")["price"].agg(["mean","median"])
```

```python
daily_price = df.groupby(["item", "date"])["price"].mean().reset_index()
```

```python
first_last = daily_price.groupby("item")["price"].agg(["first", "last"])
```

```python
first_last["percentage_change"] = (first_last["last"] / first_last["first"]) * 100
                                   
```
